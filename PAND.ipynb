Вариант 1. Базовый персептрон для логического И (AND)

Реализуйте однослойный персептрон для логической функции AND. Используйте пороговую функцию активации (step function). Веса инициализируйте нулями. Обучите на всех 4 примерах. Выведите финальные веса и предсказания.

```python
import numpy as np

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

y = np.array([0, 0, 0, 1])


weights = np.zeros(X.shape[1])
bias = 0.0
learning_rate = 0.1
epochs = 10

def step_function(x):
    return 1 if x >= 0 else 0

for epoch in range(epochs):
    for i in range(len(X)):
        linear_output = np.dot(X[i], weights) + bias
        prediction = step_function(linear_output)

        error = y[i] - prediction
        weights += learning_rate * error * X[i]
        bias += learning_rate * error


print("Финальные веса:", weights)
print(f"Финальное смещение:{bias:.1f}")

predictions = [step_function(np.dot(x, weights) + bias) for x in X]
print("Предсказания:", predictions)
```
