Вариант 3. Реализация TF-IDF с нуля
Реализуйте расчет TF-IDF без использования готовых библиотек. Даны три документа:
1. "машинное обучение интересная область"
2. "обучение с учителем ключевой аспект машинного обучения"
3. "область nlp связана с машинным обучением"

Для слова "машинное" в первом документе рассчитайте TF, IDF и TF-IDF вручную. Напишите код, проверяющий ваш расчет.

```python
import math

documents = [
    "машинное обучение интересная область",
    "обучение с учителем ключевой аспект машинного обучения",
    "область nlp связана с машинным обучением"
]

def calculate_tf(word, document):
    words = document.lower().split()
    word_count = words.count(word)
    total_words = len(words)
    return word_count / total_words

def calculate_idf(word, documents):
    N = len(documents)
    word_stem = word[:-2]
    docs_with_word = sum(1 for doc in documents 
                        if any(w.startswith(word_stem) for w in doc.lower().split()))
    return math.log(N / docs_with_word)

def simple_stem(russian_word):
    suffixes = ['ое', 'ого', 'ым', 'ом', 'ой']
    for suffix in suffixes:
        if russian_word.endswith(suffix):
            return russian_word[:-len(suffix)]
    return russian_word

def calculate_idf_stem(word, documents):
    N = len(documents)
    stem = simple_stem(word)
    docs_with_word = sum(1 for doc in documents 
                        if any(simple_stem(w) == stem for w in doc.lower().split()))
    return math.log(N / docs_with_word)

word = "машинное"
doc_index = 0
doc = documents[doc_index]

tf = calculate_tf(word, doc)
idf = calculate_idf_stem(word, documents)
tf_idf = tf * idf

print(f"Слово: {word}")
print(f"TF = {tf:.3f}")
print(f"IDF = {idf:.3f}")
print(f"TF-IDF = {tf_idf:.3f}")
```
